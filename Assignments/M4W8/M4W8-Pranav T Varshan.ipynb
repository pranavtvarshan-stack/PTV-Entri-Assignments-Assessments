1. Loading and Preprocessing (2 marks):
● Load the California Housing dataset using the fetch_california_housing function
from sklearn.
● Convert the dataset into a pandas DataFrame for easier handling. Handle
missing values (if any) and perform necessary feature scaling (e.g.,
standardization).
● Explain the preprocessing steps you performed and justify why they are
necessary for this dataset.

```python
from sklearn.datasets import fetch_california_housing
housing = fetch_california_housing()
print("Dataset loaded successfully")
```

```python
import pandas as pd
df = pd.DataFrame(
    housing.data,
    columns=housing.feature_names
)
df["MedHouseVal"] = housing.target
df.head()
 
```

### **Explanation**
The California Housing dataset is a regression dataset provided through sklearn. It contains numerical housing and geographic features that can be used to predict median house values.
 

```python

df.isnull().sum()
```

### **Explanation**
The dataset was checked for missing entries with isnull().sum(). If missing values were present, an appropriate method such as mean or median imputation could be used.
The check shows zero missing values for all columns, so no missing-value treatment is needed.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
# Separate features and target
X = df.drop("MedHouseVal", axis=1)
y = df["MedHouseVal"]
# Split dataset
X_train, X_test, y_train, y_test = train_test_split( X,y,  test_size=0.2, random_state=17)
# Standardization
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

### **Explanation**
Standardization puts the numerical features on a common scale with mean close to 0 and standard deviation close to 1.
This is particularly important for SVR because its performance can be affected by differences in feature scales.
The assignment specifically requires handling missing values and performing necessary feature scaling.

2. Regression Algorithm Implementation (5 marks):
● Implement the following regression algorithms:
○ Linear Regression
○ Decision Tree Regressor
○ Random Forest Regressor
○ Gradient Boosting Regressor
○ Support Vector Regressor (SVR)
● For each algorithm: Provide a brief explanation of how it works. Explain why it
might be suitable for this dataset.

```python
from sklearn.linear_model import LinearRegression
# Create model
linear_model = LinearRegression()
# Train model
linear_model.fit(X_train_scaled, y_train)
# Make predictions
y_pred_linear = linear_model.predict(X_test_scaled) 
print("Linear Regression completed")
```

### **Explanation**
Linear Regression is a supervised learning algorithm that assumes a linear relationship between the input features and the target variable.
It tries to find the best-fitting line/equation that minimizes the difference between actual and predicted values.
It is suitable as a baseline model because it provides a simple reference for the relationship between housing features and median house value.
 

```python
from sklearn.tree import DecisionTreeRegressor
# Create model
dt_model = DecisionTreeRegressor(random_state=17)
# Train model
dt_model.fit(X_train, y_train)
# Make predictions
y_pred_dt = dt_model.predict(X_test)
print("Decision Tree Regressor completed")
```

### **Explanation**
A Decision Tree Regressor uses a tree-like structure to make predictions. It divides the data into smaller groups using different conditions and predicts the target value at the leaf nodes.
It is suitable because it can learn non-linear relationships through feature-based splits and does not require feature scaling.
 

```python
from sklearn.ensemble import RandomForestRegressor
# Create model
rf_model = RandomForestRegressor(
    n_estimators=150,
    random_state=17
)
# Train model
rf_model.fit(X_train, y_train)
# Make predictions
y_pred_rf = rf_model.predict(X_test) 
print("Random Forest Regressor completed")
 
```

### **Explanation**
Random Forest Regressor is an ensemble learning algorithm that combines predictions from multiple decision trees.Each tree makes a prediction, and the predictions are combined to produce the final result.
It is suitable because multiple trees can capture complex non-linear patterns and the ensemble approach can reduce overfitting compared with a single tree.

```python
from sklearn.ensemble import GradientBoostingRegressor
# Create model
gb_model = GradientBoostingRegressor(
    random_state=17
)
# Train model
gb_model.fit(X_train, y_train)
# Make predictions
y_pred_gb = gb_model.predict(X_test)
print("Gradient Boosting Regressor completed")
```

### **Explanation**
Gradient Boosting Regressor builds multiple decision trees sequentially. Each new tree attempts to correct the errors made by the previous trees.
It is suitable because each successive tree focuses on correcting previous errors, helping the model learn complex relationships in the housing data.
 

```python
from sklearn.svm import SVR
# Create model
svr_model = SVR()
# Train model using scaled data
svr_model.fit(X_train_scaled, y_train)
# Make predictions
y_pred_svr = svr_model.predict(X_test_scaled)
print("SVR completed")
```

### **Explanation**
Support Vector Regressor (SVR) is a supervised learning algorithm that tries to find a function that predicts the target values while keeping prediction errors within a specified margin.
It is suitable because SVR can model non-linear relationships, and standardized features are useful because the algorithm is sensitive to feature scale.
 

3. Model Evaluation and Comparison (2 marks):
● Evaluate the performance of each algorithm using the following metrics:
○ Mean Squared Error (MSE)
○ Mean Absolute Error (MAE)
○ R-squared Score (R²)
● Compare the results of all models and identify: The best-performing algorithm
with justification and the worst-performing algorithm with reasoning.

```python
from sklearn.metrics import mean_squared_error
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score
# Store predictions
predictions = {
    "Linear Regression": y_pred_linear,
    "Decision Tree": y_pred_dt,
    "Random Forest": y_pred_rf,
    "Gradient Boosting": y_pred_gb,
    "SVR": y_pred_svr
}
# Evaluate each model
results = []
for model_name, y_pred in predictions.items():
 
    mse = mean_squared_error(y_test, y_pred)
    mae = mean_absolute_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
 
    results.append({
        "Model": model_name,
        "MSE": mse,
        "MAE": mae,
        "R2 Score": r2
    })
# Create comparison DataFrame
results_df = pd.DataFrame(results)
results_df
```

```python
# Round the results for easier reading
results_df.round(4)
```

```python
best_model = results_df.loc[
    results_df["R2 Score"].idxmax()
]
print("Best Performing Model:")
print(best_model)
 
```

```python
worst_model = results_df.loc[
    results_df["R2 Score"].idxmin()
]
print("Worst Performing Model:")
print(worst_model)
```

### **Explanation**
The `mean_squared_error`, `mean_absolute_error`, and `r2_score` functions are imported from `sklearn.metrics` to evaluate the performance of the regression models. The predictions of all five regression models are stored in a dictionary called `predictions`. A `for` loop is then used to calculate the MSE, MAE, and R² score for each model.The calculated evaluation metrics are stored in a list called `results`. Finally, the list is converted into a Pandas DataFrame named `results_df`, which makes it easier to compare the performance of all five regression models.
A lower MSE and MAE indicate better performance, while a higher R² score indicates better performance.
