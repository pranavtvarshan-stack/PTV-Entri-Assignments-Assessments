```python
import pandas as pd
import numpy as np

url = "winequality-red.csv"  
df = pd.read_csv(url, sep=';')

print("Missing values before imputation:")
print(df.isnull().sum())
df.fillna(df.mean(), inplace=True)

print("\nMissing values after imputation:")
print(df.isnull().sum())
```

```python
# Extract 'alcohol' and 'citric acid' as vectors
alcohol_vec = df['alcohol'].to_numpy()
citric_acid_vec = df['citric acid'].to_numpy()

print(f"Alcohol vector shape: {alcohol_vec.shape}")
print(f"Citric Acid vector shape: {citric_acid_vec.shape}")
print(f"First 5 elements of Alcohol: {alcohol_vec[:5]}")
print(f"First 5 elements of Citric Acid: {citric_acid_vec[:5]}")
```

```python
X = df[['alcohol', 'density']].values  

# Compute covariance matrix 
cov_matrix = np.cov(X.T)

print("Covariance Matrix:")
print(cov_matrix)
```

```python
# Eigen decomposition
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

sorted_indices = np.argsort(eigenvalues)[::-1]
top_eigenvalues = eigenvalues[sorted_indices]
top_eigenvectors = eigenvectors[:, sorted_indices]

print("Top 2 Eigenvalues:")
print(top_eigenvalues)

print("\nCorresponding Eigenvectors (Columns):")
print(top_eigenvectors)
```

```python
# Frequency count of wine quality scores
quality_counts = df['quality'].value_counts().sort_index()
print("Distribution of Wine Quality Scores:")
print(quality_counts)

most_common_quality = df['quality'].mode()[0]
print(f"\nMost common wine quality score: {most_common_quality}")
```
