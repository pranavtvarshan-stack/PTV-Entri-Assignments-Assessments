## 1. Import Required Libraries

```python
import pandas as pd
import numpy as np

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
```

## 2. Load the Breast Cancer Dataset

```python
cancer = load_breast_cancer()

X = pd.DataFrame(cancer.data, columns=cancer.feature_names)
y = pd.Series(cancer.target, name="target")

print("Number of rows and columns:", X.shape)
print("Target size:", y.shape)

X.head()
```

## 3. Check for Missing Values

The dataset is checked for missing values before model training.

```python
missing = X.isna().sum()

print("Total missing values:", missing.sum())
print("\nMissing values by column:")
print(missing)
```

### Preprocessing

The dataset contains numerical features. Median imputation is included in the model pipelines to handle possible missing numerical values. Standardization is applied to Logistic Regression, SVM, and k-NN because these methods are sensitive to feature scale. Decision Tree and Random Forest do not require scaling.

## 4. Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)
```

## 5. Logistic Regression

**How it works:** Logistic Regression estimates the probability that an observation belongs to a class.

**Why suitable:** The Breast Cancer dataset is a binary classification problem.

```python
log_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(max_iter=5000))
])

log_model.fit(X_train, y_train)
pred_log = log_model.predict(X_test)

print("Logistic Regression Accuracy:",
      accuracy_score(y_test, pred_log))
```

## 6. Decision Tree Classifier

**How it works:** A Decision Tree splits observations according to feature values and follows branches to make a classification.

**Why suitable:** It can model nonlinear relationships and does not require feature scaling.

```python
tree_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("classifier", DecisionTreeClassifier(random_state=42))
])

tree_model.fit(X_train, y_train)
pred_tree = tree_model.predict(X_test)

print("Decision Tree Accuracy:",
      accuracy_score(y_test, pred_tree))
```

## 7. Random Forest Classifier

**How it works:** Random Forest combines predictions from multiple decision trees to make a final classification.

**Why suitable:** It can work with many numerical features and model complex relationships.

```python
forest_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("classifier", RandomForestClassifier(n_estimators=100, random_state=42))
])

forest_model.fit(X_train, y_train)
pred_forest = forest_model.predict(X_test)

print("Random Forest Accuracy:",
      accuracy_score(y_test, pred_forest))
```

## 8. Support Vector Machine (SVM)

**How it works:** SVM finds a decision boundary that separates classes while maximizing the margin between them.

**Why suitable:** It can work effectively with numerical features and complex class boundaries. Scaling is important for SVM.

```python
svm_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("classifier", SVC())
])

svm_model.fit(X_train, y_train)
pred_svm = svm_model.predict(X_test)

print("SVM Accuracy:",
      accuracy_score(y_test, pred_svm))
```

## 9. k-Nearest Neighbors (k-NN)

**How it works:** k-NN classifies an observation using the classes of its nearest neighboring observations.

**Why suitable:** The dataset has numerical features, allowing distance-based classification. Scaling is important for k-NN.

```python
knn_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("classifier", KNeighborsClassifier(n_neighbors=5))
])

knn_model.fit(X_train, y_train)
pred_knn = knn_model.predict(X_test)

print("k-NN Accuracy:",
      accuracy_score(y_test, pred_knn))
```

## 10. Compare Model Performance

```python
classifiers = {
    "Logistic Regression": log_model,
    "Decision Tree": tree_model,
    "Random Forest": forest_model,
    "SVM": svm_model,
    "k-NN": knn_model
}

comparison = []

for model_name, classifier in classifiers.items():
    classifier.fit(X_train, y_train)
    predictions = classifier.predict(X_test)

    comparison.append([
        model_name,
        accuracy_score(y_test, predictions),
        precision_score(y_test, predictions),
        recall_score(y_test, predictions),
        f1_score(y_test, predictions)
    ])

comparison_df = pd.DataFrame(
    comparison,
    columns=["Model", "Accuracy", "Precision", "Recall", "F1 Score"]
)

comparison_df
```

## 11. Highest and Lowest Accuracy

If multiple models have the same highest accuracy, all tied models are reported.

```python
highest_accuracy = comparison_df["Accuracy"].max()
lowest_accuracy = comparison_df["Accuracy"].min()

highest = comparison_df.loc[
    comparison_df["Accuracy"] == highest_accuracy, "Model"
].tolist()

lowest = comparison_df.loc[
    comparison_df["Accuracy"] == lowest_accuracy, "Model"
].tolist()

print("Highest accuracy model(s):", ", ".join(highest))
print("Highest accuracy:", highest_accuracy)

print("\nLowest accuracy model(s):", ", ".join(lowest))
print("Lowest accuracy:", lowest_accuracy)
```

## 12. Conclusion

Five supervised classification algorithms were implemented using the Breast Cancer dataset: Logistic Regression, Decision Tree, Random Forest, SVM, and k-NN. Missing values were checked and preprocessing was applied using median imputation and feature scaling where appropriate. The models were compared using accuracy, precision, recall, and F1-score.

The highest and lowest accuracy models are reported from the actual test-set results above.
